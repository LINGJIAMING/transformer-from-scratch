```python
class Mylinear:
    def __init__(self, in_features , out_features):
        print("1.__init__ 被调用，初始化参数")
        self.weight = torch.randn(out_features, in_features)
        self.bias = torch.zeros(out_features)

    def forward(self,x):
        print("2. forward 被调用： 执行计算")
        return torch.matmul(x  ,self.weight.T) + self.bias

    def __call__(self,x):
        print("3. __call__ 被调用，开始处理输入")
        #这里可以做各种预处理
        if not isinstance(x,torch.Tensor):
            x = torch.tensor(x , dtype=torch.float32)
        print("预处理完成，准备调用 forward")

        result = self.forward(x)

        print("4.后处理完成，返回结果")

        return result


model = Mylinear(4,3)
x= [[1. , 2. , 3. , 4. ]]

print("\n ---使用model（x）调用——————")
out1 = model(x)

print("\n ---使用model.forward（x）调用——————")
out2 = model.forward(torch.tensor(x))



```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt

# 设置设备
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"使用设备: {device}")

# 数据预处理
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))  # MNIST数据集的均值和标准差
])

# 加载数据集
train_dataset = datasets.MNIST(
    root='./data', 
    train=True, 
    download=True, 
    transform=transform
)
test_dataset = datasets.MNIST(
    root='./data', 
    train=False, 
    download=True, 
    transform=transform
)

# 创建数据加载器
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

# 定义神经网络模型
class MNISTNet(nn.Module):
    def __init__(self):
        super(MNISTNet, self).__init__()
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.dropout1 = nn.Dropout2d(0.25)
        self.dropout2 = nn.Dropout2d(0.5)
        self.fc1 = nn.Linear(64 * 7 * 7, 128)
        self.fc2 = nn.Linear(128, 10)
    
    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = F.relu(self.conv2(x))
        x = F.max_pool2d(x, 2)
        x = self.dropout1(x)
        x = torch.flatten(x, 1)
        x = F.relu(self.fc1(x))
        x = self.dropout2(x)
        x = self.fc2(x)
        return F.log_softmax(x, dim=1)

# 初始化模型、损失函数和优化器
model = MNISTNet().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 训练函数
def train(model, device, train_loader, optimizer, epoch):
    model.train()
    train_loss = 0
    correct = 0
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)
        optimizer.zero_grad()
        output = model(data)
        loss = criterion(output, target)
        loss.backward()
        optimizer.step()
        
        train_loss += loss.item()
        pred = output.argmax(dim=1, keepdim=True)
        correct += pred.eq(target.view_as(pred)).sum().item()
        
        if batch_idx % 100 == 0:
            print(f'Train Epoch: {epoch} [{batch_idx * len(data)}/{len(train_loader.dataset)} '
                  f'({100. * batch_idx / len(train_loader):.0f}%)]\tLoss: {loss.item():.6f}')
    
    train_loss /= len(train_loader)
    accuracy = 100. * correct / len(train_loader.dataset)
    print(f'\n训练集平均损失: {train_loss:.4f}, 准确率: {correct}/{len(train_loader.dataset)} ({accuracy:.2f}%)\n')
    return train_loss, accuracy

# 测试函数
def test(model, device, test_loader):
    model.eval()
    test_loss = 0
    correct = 0
    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            test_loss += criterion(output, target).item()
            pred = output.argmax(dim=1, keepdim=True)
            correct += pred.eq(target.view_as(pred)).sum().item()
    
    test_loss /= len(test_loader)
    accuracy = 100. * correct / len(test_loader.dataset)
    print(f'测试集平均损失: {test_loss:.4f}, 准确率: {correct}/{len(test_loader.dataset)} ({accuracy:.2f}%)\n')
    return test_loss, accuracy

# 训练模型
epochs = 10
train_losses, train_accuracies = [], []
test_losses, test_accuracies = [], []

for epoch in range(1, epochs + 1):
    train_loss, train_acc = train(model, device, train_loader, optimizer, epoch)
    test_loss, test_acc = test(model, device, test_loader)
    
    train_losses.append(train_loss)
    train_accuracies.append(train_acc)
    test_losses.append(test_loss)
    test_accuracies.append(test_acc)

# 可视化训练过程
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(range(1, epochs+1), train_losses, label='训练损失')
plt.plot(range(1, epochs+1), test_losses, label='测试损失')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(range(1, epochs+1), train_accuracies, label='训练准确率')
plt.plot(range(1, epochs+1), test_accuracies, label='测试准确率')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.legend()
plt.tight_layout()
plt.show()

# 保存模型
torch.save(model.state_dict(), 'mnist_cnn.pth')
print("模型已保存为 'mnist_cnn.pth'")
```

```python
def naive_add(x, y):
    # x and y are rank-2 NumPy tensors.
    assert len(x.shape) == 2
    assert x.shape == y.shape
    # Avoids overwriting the input tensor
    x = x.copy()
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            x[i, j] += y[i, j]
    return x
```

```python
def naive_relu(x):
    # x is a rank-2 NumPy tensor.
    assert len(x.shape) == 2
    # Avoids overwriting the input tensor
    x = x.copy()
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            x[i, j] = max(x[i, j], 0)
    return x
```

```python
from torch.utils.data import DataLoader, TensorDataset
import torch.optim as optim
from keras.datasets import mnist

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# 1. 数据预处理与 Tensor 转换
# 注意：PyTorch 卷积层要求的形状是 (Batch, Channel, Height, Width)
# Keras 是 (Batch, H, W, C)，所以我们需要用 permute 调换维度
train_images_tensor = torch.from_numpy(train_images).unsqueeze(-1).permute(0, 3, 1, 2).float()/255.0 
train_labels_tensor = torch.from_numpy(train_labels).long() # 类别标签用 long 型

# 2. 封装 Dataset 和 DataLoader (对应 Keras 的 batch_size=64)
train_ds = TensorDataset(train_images_tensor, train_labels_tensor)
train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)

# 3. 定义设备、模型、优化器和损失函数
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = MyCNN().to(device)

# 对应 Keras 的 sparse_categorical_crossentropy
# PyTorch 的这个函数会自动处理 Softmax，所以模型输出层不需要加 Softmax
criterion = nn.CrossEntropyLoss() 

# 对应 optimizer="adam"
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 4. 编写训练循环 (对应 model.fit, epochs=5)
def train_one_epoch(epoch):
    model.train() # 切换到训练模式
    for batch_idx, (data, target) in enumerate(train_loader):
        # 将数据移到 GPU/CPU
        data, target = data.to(device), target.to(device)
        
        # 梯度清零 (PyTorch 必须手动做这一步)
        optimizer.zero_grad()
        
        # 前向传播
        output = model(data)
        
        # 计算损失
        loss = criterion(output, target)
        
        # 反向传播 (计算梯度)
        loss.backward()
        
        # 更新参数 (走一步优化)
        optimizer.step()

# 执行 5 个 Epoch
for epoch in range(1, 6):
    train_one_epoch(epoch)
    print(f"Epoch {epoch} finished.")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels,pooling=False):
        super(ResidualBlock, self).__init__()
        self.pooling = pooling
        
        # 1. 主路径：为了保持尺寸不变，必须设置 padding=1
        self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1)
        
        # 2. 残差路径 (Shortcut/Residual path)
        self.residual_layer = nn.Identity()
        
        # 如果需要下采样（池化）或者通道数改变，残差路径需要一个 1x1 卷积来匹配形状
        if pooling:
            # 填空：如果主路径要做 2x2 池化，残差路径为了对齐，需要设置 stride 为多少？
            self.residual_layer = nn.Conv2d(in_channels, out_channels, kernel_size=1, stride=2)
            self.pool = nn.MaxPool2d(kernel_size=2)
        elif in_channels != out_channels:
            # 如果只是通道数变了，尺寸没变
            self.residual_layer = nn.Conv2d(in_channels, out_channels, kernel_size=1)

    def forward(self, x):
        residual = self.residual_layer(x) # 拿到经过调整的残差
        
        # 主路径计算
        out = F.relu(self.conv1(x))
        out = F.relu(self.conv2(out))
        
        if self.pooling:
            out = self.pool(out) # 【填空】执行池化
            
        # 【最核心的一步】：将主路径输出与残差相加
        out = out + residual # 【填空】加什么？
        
        return F.relu(out) # 最后再过一遍激活函数
```

```python
import torch
import torch.nn as nn

class DepthwiseSeparableConv(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super(DepthwiseSeparableConv, self).__init__()
        
        # 1. Depthwise Convolution (逐通道卷积)
        # 核心：每个输入通道由一个独立的卷积核负责，不混合通道信息
        # 填空：groups 参数应该设为多少才能实现“逐通道”？
        self.depthwise = nn.Conv2d(in_channels, in_channels, kernel_size=3, 
                                   padding=1, stride=stride, groups=in_channels)
        
        # 2. Pointwise Convolution (逐点卷积)
        # 核心：使用 1x1 卷积来混合不同通道的信息，并改变输出维度
        self.pointwise = nn.Conv2d(in_channels, out_channels, kernel_size=1)
        
        self.bn = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.depthwise(x)
        x = self.pointwise(x)
        x = self.bn(x)
        return self.relu(x)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# 1. 定义深度可分离卷积模块 (SeparableConv2d)
class SeparableConv2d(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, padding=0, bias=False):
        super(SeparableConv2d, self).__init__()
        # Depthwise: groups 等于输入通道数
        self.depthwise = nn.Conv2d(in_channels, in_channels, kernel_size, 
                                   padding=padding, groups=in_channels, bias=bias) # 【填空1】
        # Pointwise: 1x1 卷积
        self.pointwise = nn.Conv2d(in_channels, out_channels, kernel_size=1, bias=bias)

    def forward(self, x):
        x = self.depthwise(x)
        x = self.pointwise(x)
        return x

# 2. 定义带残差的 Xception 块
class XceptionBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super(XceptionBlock, self).__init__()
        
        self.residual_conv = nn.Conv2d(in_channels, out_channels, kernel_size=1, stride=2, bias=False)
        
        # 主路径：BN -> ReLU -> SeparableConv -> BN -> ReLU -> SeparableConv -> MaxPool
        self.block = nn.Sequential(
            nn.BatchNorm2d(in_channels),
            nn.ReLU(),
            SeparableConv2d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(),
            SeparableConv2d(out_channels, out_channels, kernel_size=3, padding=1),
            nn.MaxPool2d(kernel_size=3, stride=2, padding=1) # 【填空2】对应 Keras 的 strides=2
        )

    def forward(self, x):
        # 计算残差路径和主路径
        residual = self.residual_conv(x)
        x = self.block(x)
        return x + residual # 【填空3】残差连接

# 3. 主模型架构
class XceptionMini(nn.Module):
    def __init__(self):
        super(XceptionMini, self).__init__()
        
        # 第一层：普通卷积（正如 Keras 注释所说，RGB 通道高度相关，不宜直接用深度可分离）
        self.init_conv = nn.Conv2d(3, 32, kernel_size=5, stride=1, bias=False)
        
        # 串联 Xception 块：32 -> 64 -> 128 -> 256 -> 512
        self.block1 = XceptionBlock(32, 64)
        self.block2 = XceptionBlock(64, 128)
        self.block3 = XceptionBlock(128, 256)
        self.block4 = XceptionBlock(256, 512)
        
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.dropout = nn.Dropout(0.5)
        self.fc = nn.Linear(512, 1)

    def forward(self, x):
        x = self.init_conv(x)
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        
        x = self.gap(x)
        x = torch.flatten(x, 1)
        x = self.dropout(x)
        return self.fc(x)
```

```python
import torch
import torch.nn.functional as F

def basic_attention(Q, K, V):
    # Q: Query, K: Key, V: Value
    # 1. 计算注意力得分：Q 乘以 K 的转置
    # 2. 缩放平滑：除以根号下 d_k (防止梯度消失)
    d_k = Q.size(-1)
    scores = torch.matmul(Q, K.transpose(-2, -1)) / torch.sqrt(torch.tensor(d_k,dtype = torch.float32)) # 【填空】
    
    # 3. 归一化：通过哪个函数变成概率分布？
    attn_weights = F.softmax(scores, dim=-1) # 【填空】
    
    # 4. 加权求和：得分乘以 Value
    output = torch.matmul(attn_weights, V) # 【填空】
    
    return output
```
