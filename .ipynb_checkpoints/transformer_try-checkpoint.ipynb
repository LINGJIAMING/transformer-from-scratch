```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class TinyAttention(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        # Q, K, V 是 Transformer 的三个灵魂矩阵
        self.q = nn.Linear(d_model, d_model)
        self.k = nn.Linear(d_model, d_model)
        self.v = nn.Linear(d_model, d_model)

    def forward(self, x):
        # 1. 计算 Q, K, V
        Q = self.q(x)
        K = self.k(x)
        V = self.v(x)
        
        # 2. 计算注意力得分 (Score = Q * K^T)
        # d_k 是缩放因子，防止数值过大
        d_k = Q.size(-1)
        scores = torch.matmul(Q, K.transpose(-2, -1)) / torch.sqrt(torch.tensor(d_k))
        
        # 3. Softmax 归一化得分
        attention_weights = F.softmax(scores, dim=-1)
        
        # 4. 加权求和得到输出
        output = torch.matmul(attention_weights, V)
        return output

# --- 模拟数据流 ---
d_model = 4  # 假设每个数字被表示为一个 4 维向量
seq_len = 3  # 假设输入序列长度是 3 (比如 "我 爱 你")
batch_size = 1 # 每次处理 1 组数据

# 构造一个随机输入 [batch, seq_len, d_model]
input_data = torch.randn(batch_size, seq_len, d_model)

# 实例化并运行
model = TinyAttention(d_model)
output = model(input_data)

print(f"输入形状: {input_data.shape}")
print(f"输出形状: {output.shape}")
print("\n输出内容（这就是模型‘观察’序列后的特征）:\n", output)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class TinyAttention(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        #embedding层
        self.embedding = nn.embedding(input_size,d_model)
        #位置编码矩阵
        self.pe = pe()
        #
        # Q, K, V 是 Transformer 的三个灵魂矩阵
        self.q = nn.Linear(d_model, d_model)
        self.k = nn.Linear(d_model, d_model)
        self.v = nn.Linear(d_model, d_model)


        #正则化
        self.norm = nn.LayerNorm(d_model)

        #
        self.ffn_1 = nn.Linear(d_model,d_ff)
        self.ffn_2 = nn.Linear(d_ff,d_model)
        



    def pe(self, x, d_model):
    """
    x: 输入张量，形状通常为 (seq_len, embedding_dim)
    d_model: 词嵌入的维度
    """
        n = len(x)  # 序列长度 seq_len
        
        # 1. 初始化位置编码矩阵
        pe = np.zeros((n, d_model))
        
        # 2. 生成位置索引 pos: [0, 1, 2, ..., n-1]
        # 需要将其重塑为 (n, 1) 以便进行广播计算
        pos = np.arange(n).reshape(-1, 1)
        
        # 3. 计算分母项 (div_term)
        # 只需计算 d_model/2 个值，因为奇偶列共用
        i = np.arange(0, d_model, 2)
        div_term = np.exp(i * -(np.log(10000.0) / d_model))
        
        # 4. 填充矩阵
        # pe[:, 0::2] 处理偶数索引 (sin)
        # pe[:, 1::2] 处理奇数索引 (cos)
        pe[:, 0::2] = np.sin(pos * div_term)
        pe[:, 1::2] = np.cos(pos * div_term)
        
        return pe

    def forward(self, x):
        # 1.词嵌入
        x = self.embedding(x,512)
        # 2.加入位置编码矩阵
        x = x + pe(x,512)
        
        # 3. 计算 Q, K, V(简单的单头注意力机制)
        Q = self.q(x)
        K = self.k(x)
        V = self.v(x)
        
        # 2. 计算注意力得分 (Score = Q * K^T)
        # d_k 是缩放因子，防止数值过大
        d_k = Q.size(-1)
        scores = torch.matmul(Q, K.transpose(-2, -1)) / torch.sqrt(torch.tensor(d_k))
        
        # 3. Softmax 归一化得分
        attention_weights = F.softmax(scores, dim=-1)
        
        # 4. 加权求和得到输出
        output = torch.matmul(attention_weights, V)

        # 5. 加入残差以及正则化
        output = self.norm(x + output)

        # 6. 前馈网络
        output = output + self.norm(self.ffn_2(max(0,self.ffn_1(output))))

        return output
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np

class TinyAttention(nn.Module):
    def __init__(self, input_size, d_model, d_ff,h):
        super().__init__()
        self.d_model = d_model
        
        # 1. Embedding 层：注意类名大写
        self.embedding = nn.Embedding(input_size, d_model, padding_idx=0)

        self.h = h
        # 2. Q, K, V 线性层
        self.d_k = d_model // h
        
        self.q = nn.Linear(d_model, d_model)
        self.k = nn.Linear(d_model, d_model)
        self.v = nn.Linear(d_model, d_model)
        
        self.out_proj = nn.Linear(d_model, d_model)

        # 3. 正则化
        self.norm = nn.LayerNorm(d_model)

        # 4. 前馈网络 (FFN)
        self.ffn_1 = nn.Linear(d_model, d_ff)
        self.ffn_2 = nn.Linear(d_ff, d_model)

    def get_pe(self, seq_len, d_model):
        """生成位置编码矩阵"""
        pe = np.zeros((seq_len, d_model))
        pos = np.arange(seq_len).reshape(-1, 1)
        i = np.arange(0, d_model, 2)
        div_term = np.exp(i * -(np.log(10000.0) / d_model))
        
        pe[:, 0::2] = np.sin(pos * div_term)
        pe[:, 1::2] = np.cos(pos * div_term)
        
        # 转换为 torch 张量并移动到与输入相同的设备上
        return torch.tensor(pe, dtype=torch.float32)

    def forward(self, x):
        # x 形状: (batch_size, seq_len)
        
        # 1. 词嵌入
        # x 结果形状: (batch_size, seq_len, d_model) 2*10*512 --> 2* 10*8*64
        x = self.embedding(x)
        # x 形状: (batch_size, seq_len)
        batch_size = x.size(0)  # 获取真正的 batch_size (例如 2)
        seq_len = x.size(1)     # 获取序列长度 (例如 10)

        
        # 2. 加入位置编码矩阵
        # 动态获取当前 batch 的序列长度
        pe = self.get_pe(x.size(1), self.d_model).to(x.device)
        x = x + pe
        
        # 3. 计算 Q, K, V
        Q = self.q(x).view(batch_size, -1, self.h, self.d_k).transpose(1, 2)
        K = self.k(x).view(batch_size, -1, self.h, self.d_k).transpose(1, 2)
        V = self.v(x).view(batch_size, -1, self.h, self.d_k).transpose(1, 2)
        
        # 4. 计算注意力得分
        
        # scores 形状: (batch_size, seq_len, seq_len)
        scores = torch.matmul(Q, K.transpose(-2, -1)) / torch.sqrt(torch.tensor(self.d_k, dtype=torch.float32))
        
        
        # 5. Softmax
        attention_weights = F.softmax(scores, dim=-1)
        
        # 6. 加权求和得到输出
        attn_output = torch.matmul(attention_weights, V)
        attn_output = attn_output.transpose(1, 2)
        attn_output = attn_output.contiguous().view(batch_size, seq_len, -1)

        attn_output = self.out_proj(attn_output)
        
        
        
        # 7. 残差连接 + LayerNorm (Attention 部分)
        x = self.norm(x + attn_output)

        # 8. 前馈网络 (FFN) + 残差连接
        # 使用 F.relu 替代 max(0, x)
        ffn_output = self.ffn_2(F.relu(self.ffn_1(x)))
        output = self.norm(x + ffn_output)

        return output

# --- 测试代码 ---
model = TinyAttention(input_size=100, d_model=512, d_ff=2048,h=8)
test_input = torch.randint(0, 100, (2, 10)) # batch_size=2, seq_len=10
output = model(test_input)
print(output.shape) # 预期输出: torch.Size([2, 10, 512])
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import math

class MultiHeadAttention(nn.Module):
    """通用的多头注意力模块"""
    def __init__(self, d_model, h, dropout=0.1):
        super().__init__()
        self.h = h
        self.d_k = d_model // h
        
        self.q_linear = nn.Linear(d_model, d_model)
        self.k_linear = nn.Linear(d_model, d_model)
        self.v_linear = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)
        
        self.dropout = nn.Dropout(dropout)

    def forward(self, q, k, v, mask=None):
        batch_size = q.size(0)
        
        # 1. 线性变换并拆分头
        Q = self.q_linear(q).view(batch_size, -1, self.h, self.d_k).transpose(1, 2)
        K = self.k_linear(k).view(batch_size, -1, self.h, self.d_k).transpose(1, 2)
        V = self.v_linear(v).view(batch_size, -1, self.h, self.d_k).transpose(1, 2)

        # 2. 计算注意力得分
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.d_k)
        
        # 3. 应用 Mask (针对 Decoder 的关键步骤)
        if mask is not None:
            scores = scores.masked_fill(mask == 0, -1e9)
            
        weights = F.softmax(scores, dim=-1)
        weights = self.dropout(weights) # 细节：对注意力权重做 Dropout

        # 4. 合并头
        attn_output = torch.matmul(weights, V)
        attn_output = attn_output.transpose(1, 2).contiguous().view(batch_size, -1, self.h * self.d_k)
        
        return self.out_proj(attn_output)

class EncoderLayer(nn.Module):
    def __init__(self, d_model, h, d_ff, dropout=0.1):
        super().__init__()
        self.self_attn = MultiHeadAttention(d_model, h, dropout)
        self.feed_forward = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(d_ff, d_model)
        )
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # Attention + Residual + Norm
        attn_out = self.self_attn(x, x, x)
        x = self.norm1(x + self.dropout(attn_out))
        
        # FFN + Residual + Norm
        ffn_out = self.feed_forward(x)
        x = self.norm2(x + self.dropout(ffn_out))
        return x

class DecoderLayer(nn.Module):
    def __init__(self, d_model, h, d_ff, dropout=0.1):
        super().__init__()
        self.self_attn = MultiHeadAttention(d_model, h, dropout)
        self.cross_attn = MultiHeadAttention(d_model, h, dropout) # 接收 Encoder 信息
        
        self.feed_forward = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(d_ff, d_model)
        )
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.norm3 = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, enc_output, src_mask=None, trg_mask=None):
        # 1. Masked Self-Attention (只看自己之前的内容)
        x = self.norm1(x + self.dropout(self.self_attn(x, x, x, trg_mask)))
        
        # 2. Cross-Attention (Q来自自己, K,V来自Encoder)
        # 这步是 Decoder 翻译/生成的依据
        x = self.norm2(x + self.dropout(self.cross_attn(x, enc_output, enc_output, src_mask)))
        
        # 3. FFN
        x = self.norm3(x + self.dropout(self.feed_forward(x)))
        return x
```
